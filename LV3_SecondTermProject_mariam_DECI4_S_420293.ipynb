# Task 1: Build the Data Collection Pipeline

## Step 1: Collect API Data

```python
import requests
import pandas as pd

url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)

data = response.json()

print(response.status_code)
```

## Step 2: Create the Dataset

```python
df = pd.DataFrame(data["items"])
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns
```

## Step 3: Select Required Columns

```python
df = df[[ "name", "owner", "language", "stargazers_count", "forks_count", "watchers_count",
         "open_issues_count", "created_at", "updated_at", "license" ]]

df.head()
```

## Step 4: Extract Required Values

```python
df["owner"] = df["owner"].apply(lambda x: x["login"])
df["license"] = df["license"].apply(lambda x: x["name"] if x else None)

df.head()
```

## Step 5: Handle Missing Values

```python
df.isnull().sum()
```

```python
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("Unknown")
```

```python
df.isnull().sum()
```

## Step 6: Handle Duplicate Records

```python
df.duplicated().sum()
```

## Step 7: Prepare the Dataset

**first : the created_at and updated_at columns will converted into datatime**

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])
```

```python
df.dtypes
```

**second : several column names will be shortened**

```python
df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})

df.columns
```

## Step 8: Save and Verify the Dataset

```python
df.to_csv("github_projects.csv", index=False)
```

# Task 2: Store and Analyze the Data

## Step 1: Create the SQLite Database

```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3

df = pd.read_csv("github_projects.csv")
connection = sqlite3.connect("github_projects.db")

df.to_sql("Repositories", connection, if_exists="replace", index=False)

!pip install ipython-sql==0.5.0 prettytable==3.9.0

%load_ext sql
%config SqlMagic.style = 'PLAIN_COLUMNS'
%sql sqlite:///github_projects.db
```

```python
%%sql
SELECT * FROM Repositories
LIMIT 5;
```

## Step 2: Filtering & Searching Queries

```python
%%sql

SELECT name , stars
FROM Repositories
WHERE stars > 10000;
```

```python
%%sql

SELECT name
FROM Repositories
WHERE name LIKE '%Machine%';
```

## Step 3: Logical Operators

**1. AND**

```python
%%sql

SELECT *
FROM Repositories
WHERE stars > 10000
AND language = 'Python';
```

**2. OR**

```python
%%sql

SELECT *
FROM Repositories
WHERE language = 'Python'
OR language = 'JavaScript';
```

**3. NOT**

```python
%%sql

SELECT *
FROM Repositories
WHERE NOT language = 'Python';
```

## Step 4: Sorting & Limiting Queries

**first : Sort the repositories according to the number of stars**

```python
%%sql

SELECT name , stars
FROM Repositories
ORDER BY stars DESC;
```

**second :  Top 10 repositories with the highest number of stars**

```python
%%sql

SELECT name, stars
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
```

## Step 5: Aggregate Functions

**first : the total number of repositories.**

```python
%%sql

SELECT COUNT(*) AS total_repositories
FROM Repositories;
```

**second : the average number of stars.**

```python
%%sql

SELECT AVG(stars) AS average_stars
FROM Repositories;
```

## Step 6: Grouping Analysis

```python
%%sql

SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5;
```

## Step 7: Create Python Visualizations

```python
top10 = df.sort_values(by='stars', ascending=False).head(10)
plt.bar(top10["name"],top10["stars"], color='purple')

plt.title("Top 10 GitHub Repositories by Stars")
plt.xlabel("Repository Name")
plt.ylabel("Stars")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df['created_date'] = pd.to_datetime(df['created_date'])

yearly_repositories = (df.groupby(df['created_date'].dt.year).size().reset_index(name='repository_count'))

plt.figure(figsize=(10,6))
plt.plot(yearly_repositories['created_date'],
    yearly_repositories['repository_count'],
    marker='o')

plt.xlabel('Year')
plt.ylabel('Number of Repositories')
plt.title('Repository Creation Trends Over Time')
plt.show()
```

## Step 8: Interpret Results

The analysis shows that the transformers have significantly higher numbers of stars compared with the rest of the dataset. This indicates that a small group of repositories attracts a large share of user interest

The repository creation trend shows an overall increase at 2018, indicating growing activity and interest in GitHub projects
